```python
import random

random.seed(42) #using 42 so we all have the same numbers

def generate_inventory(num_categories=5, num_products_per_category=10): #creating 
    categories = ["Hand soap", "Moisturizer", "Sunscreen", "Body oil", "Shampoo"]
    inventory = {}
    for category in categories:
        inventory[category] = []
        for _ in range(num_products_per_category):
            product = {
                'Quantity': random.randint(10, 100),
                'Unit Price': round(random.uniform(5.0, 500.0), 2),
                'Expiration Date': random.randint(1, 365),
                'Advanced Ordered': random.randint(0, 8)
            } #this give all the information generated between certain parameters like (10,100) or (0,8)
            inventory[category].append(product)
    return inventory  # <- dedented, now outside the for loop
my_inventory = generate_inventory()
print(my_inventory.items()) 
```

```python
for category, products in my_inventory.items() :
    print(f'{category}:{products[0]}') 
```

```python
total_count = 0

for category, products in my_inventory.items():
    print(f" {category} : {len(products)}")
    total_count += len(products) 
print("Total products:", total_count) 
```

```python
all_expirations = []  # empty list to collect every expiration date we find

for category, products in my_inventory.items():
    for product in products:                      # loop through each of the 10 products in this category
        all_expirations.append(product['Expiration Date'])  # pull out just this one field

print("Min expiration:", min(all_expirations))
print("Max expiration:", max(all_expirations)) 
```

```python
qualifying_products = []

for category, products in my_inventory.items():
    for product in products:
        if product['Unit Price'] > 200 and product['Expiration Date'] < 30 and product['Quantity'] > product['Advanced Ordered']:
            qualifying_products.append({
                'Category': category,
                'Quantity': product['Quantity'],
                'Unit Price': product['Unit Price'],
                'Expiration Date': product['Expiration Date'],
                'Advanced Ordered': product['Advanced Ordered'],
                'Inventory Value': product['Quantity'] * product['Unit Price']
            })

print(qualifying_products)
```
