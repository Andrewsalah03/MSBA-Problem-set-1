# Part A

```python
import random

random.seed(42) #using 42 so we all have the same numbers

# Fixes I made to the original code:
# 1. changed the curly quotes to normal quotes
# 2. took out the word "and" from the category list and added the missing quote on Shampoo
# 3. changed "return inventories" to "return inventory" since inventories doesn't exist
# 4. changed expiration from (-2, 365) to (1, 365) because expired items (0 or less) get removed
# 5. fixed the indentation

def generate_inventory(num_categories=5, num_products_per_category=10): #creating the inventory
    categories = ["Hand soap", "Moisturizer", "Sunscreen", "Body oil", "Shampoo"]
    inventory = {}
    for category in categories:
        inventory[category] = []
        for _ in range(num_products_per_category):
            product = {
                'Quantity': random.randint(10, 100),
                'Unit Price': round(random.uniform(5.0, 500.0), 2),
                'Expiration Date': random.randint(1, 365),
                'Advanced Ordered': random.randint(0, 8)
            } #this give all the information generated between certain parameters like (10,100) or (0,8)
            inventory[category].append(product)
    return inventory
my_inventory = generate_inventory()
print(my_inventory.items())
```

```python
total_count = 0

for category, products in my_inventory.items():
    print(f" {category} : {len(products)}") #number of products in each category
    total_count += len(products) #adding each category to the total
print("Total products:", total_count)
```

```python
all_expirations = []  # empty list to collect every expiration date we find

for category, products in my_inventory.items():
    for product in products:                      # loop through each of the 10 products in this category
        all_expirations.append(product['Expiration Date'])  # pull out just this one field

print("Min expiration:", min(all_expirations))
print("Max expiration:", max(all_expirations))
```

```python
for category, products in my_inventory.items() :
    print(f'{category}:{products[0]}') #products[0] is the first product in each category
```

# Part B

```python
qualifying_products = []

for category, products in my_inventory.items():
    for product in products:
        # checking all 3 conditions: price over 200, less than 30 days left, quantity more than advanced ordered
        if product['Unit Price'] > 200 and product['Expiration Date'] < 30 and product['Quantity'] > product['Advanced Ordered']:
            qualifying_products.append({
                'Category': category,
                'Quantity': product['Quantity'],
                'Unit Price': product['Unit Price'],
                'Expiration Date': product['Expiration Date'],
                'Advanced Ordered': product['Advanced Ordered'],
                'Inventory Value': round(product['Quantity'] * product['Unit Price'], 2) #quantity x unit price
            })

for product in qualifying_products:
    print(product) #printing each qualifying product on its own line
```

```python
print("Total qualifying products:", len(qualifying_products))

total_value = 0
for product in qualifying_products:
    total_value += product['Inventory Value'] #adding up every inventory value
print("Aggregate inventory value:", round(total_value, 2))
```

```python
total_quantity = 0
total_advanced = 0

for product in qualifying_products:
    total_quantity += product['Quantity']
    total_advanced += product['Advanced Ordered']

avg_quantity = total_quantity / len(qualifying_products)
avg_advanced = total_advanced / len(qualifying_products)

if avg_advanced == 0: #can't divide by zero so we check first
    print("Average Advanced Ordered is 0, can't calculate the ratio")
else:
    print("Average Quantity / Average Advanced Ordered:", round(avg_quantity / avg_advanced, 2))
```

```python
category_totals = {} #will hold the total inventory value for each category

for product in qualifying_products:
    category = product['Category']
    if category not in category_totals:
        category_totals[category] = 0
    category_totals[category] = round(category_totals[category] + product['Inventory Value'], 2) #round so it doesn't print long decimals

print(category_totals)

top_category = ""
top_value = 0
for category, value in category_totals.items():
    if value > top_value: #keep whichever category has the biggest value
        top_value = value
        top_category = category

print("Largest category:", top_category, round(top_value, 2))
```

# Part C

```python
max_price = 0
for product in qualifying_products:
    if product['Unit Price'] > max_price: #finding the highest unit price
        max_price = product['Unit Price']

for product in qualifying_products:
    product['Normalized Price'] = round(product['Unit Price'] / max_price, 4) #new field, original price stays the same

most_expensive = qualifying_products[0]
least_expensive = qualifying_products[0]
for product in qualifying_products:
    if product['Unit Price'] > most_expensive['Unit Price']:
        most_expensive = product
    if product['Unit Price'] < least_expensive['Unit Price']:
        least_expensive = product

print("Most expensive unit price:", most_expensive['Unit Price'])
print("Least expensive unit price:", least_expensive['Unit Price'])
print("Least expensive normalized price:", least_expensive['Normalized Price'])
```

```python
def get_normalized(product): #tells sorted() to sort by the normalized price
    return product['Normalized Price']

sorted_products = sorted(qualifying_products, key=get_normalized, reverse=True) #reverse=True means highest first

for product in sorted_products:
    print(product)
```

A normalized Unit Price close to 1 means that product is priced almost as high as the most expensive product on the list. These are the items where SD Retailing loses the most money per unit if they expire before being sold, so they should be dealt with first, like putting them on sale.

# Part D

```python
remaining_products = []
for product in qualifying_products:
    if product != most_expensive: #leave out the most expensive product
        remaining_products.append(product)

new_totals = {}
for product in remaining_products:
    category = product['Category']
    if category not in new_totals:
        new_totals[category] = 0
    new_totals[category] = round(new_totals[category] + product['Inventory Value'], 2)

print(new_totals)

new_top_category = ""
new_top_value = 0
for category, value in new_totals.items():
    if value > new_top_value:
        new_top_value = value
        new_top_category = category

print("Old largest category:", top_category)
print("New largest category:", new_top_category)

if new_top_category == top_category:
    print("Removing the most expensive product does not change the largest category")
else:
    print("Removing the most expensive product changes the largest category")
```
